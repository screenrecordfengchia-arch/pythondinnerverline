## 題目說明

餐廳給的餐費 `" NT$1,200 "` 和人數 `"4"` 都是文字，不能直接計算。本程式先整理名稱、清理餐費並轉換型別，再依規則計算：

1. 服務費 = 原始餐費 × 10%
2. 應付總額 = 原始餐費 + 服務費 − 折價券
3. 每人費用 = 應付總額 ÷ 人數
4. 判斷每人費用是否大於每人預算 300 元（結果為布林值）

**手算預測（測試一）**：1200 × 0.10 = 120；1200 + 120 − 100 = 1220；1220 ÷ 4 = 305 > 300 → `True`

## 測試一：原始資料（4 人、折價券 100 元）

```python
# ===== 測試一：原始資料 =====
# 1. 原始資料（保留原始文字，不直接修改）
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 2. 整理聚餐名稱：去除頭尾空白，再轉小寫
clean_name = event_name.strip().lower()

# 3. 整理餐費：移除 NT$、逗號、空白後轉成 float；人數 "4" 轉成 int
meal_text = meal_amount.replace("NT$", "").replace(",", "").replace(" ", "")
meal_price = float(meal_text)
people_count = int(people)

# 4. 分步計算：服務費用「原始餐費」計算，之後才扣折價券
service_fee = meal_price * service_rate
total_due = meal_price + service_fee - discount
per_person = total_due / people_count

# 5. 比較運算：每人費用 > 每人預算 → True（等於預算不算超過）
over_budget = per_person > budget_per_person

# 6. 輸出摘要（f-string 的 :.2f 代表顯示到小數點後兩位）
print(f"聚餐名稱：{clean_name}")
print(f"原始餐費：{meal_price:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_due:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{over_budget}")
```
