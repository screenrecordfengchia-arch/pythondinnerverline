## 題目說明

整理「桌遊體驗」與「電影欣賞」兩場活動的報名文字：統一小寫、移除空白後拆成 list，再用 set 去除重複報名、比較兩場的參加者，最後依各場**去重後**的人數計算費用（桌遊 120 元、電影 80 元）。

| 型別 | 本題用途 |
|---|---|
| `tuple` | `activity_names`：固定的兩個活動名稱 |
| `str` | `board_raw`、`movie_raw`：原始報名文字 |
| `list` | 整理後保留重複的名單、排序後的名單、前兩位預覽 |
| `set` | 去重，並做交集、聯集、差集 |
| `dict` | `unit_fees` 查費用；`summary` 保存人數與金額 |

## 測試一：原始資料

輸入 `movie_raw = " ben, Dora, ERIC, ben "`

預測：桌遊 {amy, ben, cora} 3 人、電影 {ben, dora, eric} 3 人，共同 ben 1 人；總不重複 5 人、報名 6 人次；3×120 + 3×80 = 600 元

```python
# ===== 測試一：原始資料 =====
# 1. 原始資料
# tuple：保存固定的兩個活動名稱
activity_names = ("桌遊體驗", "電影欣賞")
# str：原始文字名單（保留原值，不覆蓋）
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
# dict：用活動名稱查詢每人費用
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 2. 用拆包從 tuple 取出兩個活動名稱
board_name, movie_name = activity_names

# 3. 清理文字：移除半形空白 → 轉小寫 → 用逗號拆成 list（保留順序與重複）
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 4. 轉成 set 去重
board_set = set(board_list)
movie_set = set(movie_list)

# 5. 集合運算比較兩份名單
both_set = board_set & movie_set          # 交集：兩場都參加
all_set = board_set | movie_set           # 聯集：至少參加一場
board_only_set = board_set - movie_set    # 差集：只參加桌遊
movie_only_set = movie_set - board_set    # 差集（反方向）：只參加電影

# 6. 排序成 list 才能顯示與切片
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
both_sorted = sorted(both_set)
board_only_sorted = sorted(board_only_set)
movie_only_sorted = sorted(movie_only_set)
all_sorted = sorted(all_set)
notify_preview = all_sorted[:2]           # 切片：通知名單前兩位

# 7. dict 保存摘要：人數用 len() 算「去重後」的 set，費用從 unit_fees 查詢
summary = {
    "board_count": len(board_set),
    "movie_count": len(movie_set),
    "both_count": len(both_set),
    "unique_count": len(all_set),
    "signup_times": len(board_set) + len(movie_set),
    "board_income": len(board_set) * unit_fees[board_name],
    "movie_income": len(movie_set) * unit_fees[movie_name],
}
summary["total_income"] = summary["board_income"] + summary["movie_income"]

# 8. 輸出摘要（:.2f 代表顯示到小數點後兩位）
print("HW09 活動報名摘要")
print(f"活動：{board_name}，每人 {unit_fees[board_name]:.2f} 元")
print(f"活動：{movie_name}，每人 {unit_fees[movie_name]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{both_sorted}")
print(f"共同人數：{summary['both_count']} 人")
print(f"只參加桌遊者：{board_only_sorted}")
print(f"只參加電影者：{movie_only_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['unique_count']} 人")
print(f"報名人次：{summary['signup_times']} 人次")
print(f"通知名單前兩位：{notify_preview}")
print(f"桌遊應收：{summary['board_income']:.2f} 元")
print(f"電影應收：{summary['movie_income']:.2f} 元")
print(f"總應收：{summary['total_income']:.2f} 元")
```
